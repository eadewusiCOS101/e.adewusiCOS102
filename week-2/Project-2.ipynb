```python
#Quadratic, cubic, and quartic solutions.
import math

def floating_inputs(prompt):
    while True:
        try:
            a = float(input(prompt))
            return a
        except ValueError:
            print("Invalid input received")

def cubic_solving(a,b,c,d):
    if a == 0:
        print("Invalid, a cannot be 0.")
        return
    b /= a; c /= a; d /= a
    p = c - (b**2) / 3
    q = (2 * b**3) / 27 - (b * c) / 3 + d

    discriminant = (q / 2)**2 + (p / 3)**3
    offset = b / 3 

    if discriminant > 0:
        sqrt_disc = math.sqrt(discriminant)
        u_val = -q/2 + sqrt_disc
        v_val = -q/2 - sqrt_disc
        u = math.copysign(abs(u_val)**(1/3), u_val)
        v = math.copysign(abs(v_val)**(1/3), v_val)
        x1 = u + v - offset
        real = -(u + v) / 2 - offset
        imag =  (u - v) * math.sqrt(3) / 2
        print(f"One real root:   x1 = {x1:.2f}")
        print(f"Complex roots:   x2 = {real:.2f} + {imag:.2f}i")
        print(f"                 x3 = {real:.2f} - {imag:.2f}i")

    elif discriminant == 0: 
        u = math.copysign(abs(-q/2)**(1/3), -q/2)
        x1 = 2 * u - offset
        x2 = -u - offset
        print(f"Repeated roots:  x1 = {x1:.2f},  x2 = x3 = {x2:.2f}")

    else:                    
        r = math.sqrt((-p/3)**3)
        theta = math.acos(max(-1, min(1, -q / (2 * r))))
        r_cbrt = 2 * math.copysign(abs(r)**(1/3), r)
        x1 = r_cbrt * math.cos(theta / 3)             - offset
        x2 = r_cbrt * math.cos((theta + 2*math.pi)/3) - offset
        x3 = r_cbrt * math.cos((theta + 4*math.pi)/3) - offset
        print(f"Three real roots:  x1 = {x1:.2f},  x2 = {x2:.2f},  x3 = {x3:.2f}")

def quartic_solving(a,b,c,d,e):
    if a == 0:
        print("Not a quartic equation (A cannot be 0).")
        return

    b /= a; c /= a; d /= a; e /= a

    p = c - 3*b**2/8
    q = b**3/8 - b*c/2 + d
    r = -3*b**4/256 + b**2*c/16 - b*d/4 + e

    offset = b / 4

    if abs(q) < 1e-10:     
        disc = p**2 - 4*r
        if disc < 0:
            print("All roots are complex (no real roots).")
            return
        t1 = (-p + math.sqrt(disc)) / 2
        t2 = (-p - math.sqrt(disc)) / 2
        roots = []
        for t in (t1, t2):
            if t >= 0:
                roots += [math.sqrt(t) - offset, -math.sqrt(t) - offset]
        if roots:
            for i, x in enumerate(roots, 1):
                print(f"x{i} = {x:.4f}")
        else:
            print("All roots are complex.")
        return

    cubic_solving(8, 8*p, 2*p**2 - 8*r, -(q**2))

    A3, B3, C3, D3 = 8, 8*p, 2*p**2 - 8*r, -(q**2)
    b3 = B3/A3; c3 = C3/A3; d3 = D3/A3
    pp = c3 - b3**2/3
    qq = 2*b3**3/27 - b3*c3/3 + d3
    disc3 = (qq/2)**2 + (pp/3)**3
    off3 = b3/3
    if disc3 >= 0:
        sq = math.sqrt(disc3)
        u_v = -qq/2 + sq; v_v = -qq/2 - sq
        u = math.copysign(abs(u_v)**(1/3), u_v)
        v = math.copysign(abs(v_v)**(1/3), v_v)
        m = u + v - off3
    else:
        rr = math.sqrt((-pp/3)**3)
        theta = math.acos(max(-1, min(1, -qq/(2*rr))))
        m = 2*math.copysign(abs(rr)**(1/3), rr)*math.cos(theta/3) - off3

    sqrt_2m = math.sqrt(max(0, 2*m))
    roots = []
    for sign in (+1, -1):
        A2 = 1
        B2 = sign * sqrt_2m
        C2 = p + 2*m + sign * (-q / (sqrt_2m + 1e-14))
        disc2 = B2**2 - 4*A2*C2
        if disc2 >= 0:
            roots.append((-B2 + math.sqrt(disc2)) / 2 - offset)
            roots.append((-B2 - math.sqrt(disc2)) / 2 - offset)

    if roots:
        for i, x in enumerate(roots, 1):
            print(f"x{i} = {x:.4f}")
    else:
        print("No real roots found.")

    

print("Welcome to the polynomial equation evaluation script!")
while True:
    process = input("What is the highest degree of the equation you want to solve:\n1)x²\n2)x³\n3)x⁴\n\n(Use keyword 'quit' to end the program)\n")
    if process == "1":
        print("Provided that the equation is in the form ax² + bx + c = 0")
        a = floating_inputs("What is a?")
        if a == 0:
            print("Invalid, a cannot be 0.")
            break
        b = floating_inputs("What is b?")
        c = floating_inputs("What is c?")
        d = (b**2) - (4 * a * c)

        if d >= 0:
            r1 = (-b + (d ** 0.5)) / (2 * a)
            r2 = (-b - (d ** 0.5)) / (2 * a)
            print(f"hence the roots of the quadratic equation {a}x² + {b}x + {c} = 0 are:\n {r1:.2f} and {r2:.2f}")
            if r1 == r2:
                print(f"\nTherfore, the root of the equation is {r1} twice.")
        else:
            real = -b / (2 * a)
            imag = (math.sqrt(-d)) / (2 * a)
            print(f"The complex roots are\n{real:.2f} + {imag:.2f}i and \n{real:.2f} - {imag:.2f}i")
            
    elif process == "2":
        print("Provided that the equation is in the form Ax³ + Bx² + Cx + D = 0")
        a = floating_inputs("What is a?")
        b = floating_inputs("What is b?")
        c = floating_inputs("What is c?")
        d = floating_inputs("What is d?")
        cubic_solving(a,b,c,d)

    elif process == "3":
        print("Provided that the equation is in the form Ax³ + Bx² + Cx + D = 0")
        a = floating_inputs("What is a?")
        b = floating_inputs("What is b?")
        c = floating_inputs("What is c?")
        d = floating_inputs("What is d?")
        e = floating_inputs("What is e?")
        quartic_solving(a,b,c,d,e)

    elif process.lower() == "quit":
        print("Shutting down the program.")
        break

    else:
        print("Invalid process.")
```
