```python
import pandas as pd

data = pd.read_csv("SIS.csv")

class Student_sorting:
    def __init__(self, file, data):
        self.file = file
        self.data = data

    def age_sorting(self):
        pirates = []
        yankees = []
        bulls = []
        for index, row in self.data.iterrows():
            age =  row["Age"]
            if age > 14 and age < 18:
                pirates.append(row)
            elif age > 18 and age < 22:
                yankees.append(row)
            elif age > 22 and age < 25:
                bulls.append(row)
            else:
                error = Encapsulation_error_handling(age, row)
                print(error.error_getter())
        Pirates_df = pd.DataFrame(pirates)
        Pirates_df.to_csv("The Pirates.csv", index = False)
        Yankees_df = pd.DataFrame(yankees)
        Yankees_df.to_csv("The Yankees.csv", index = False)
        Bulls_df = pd.DataFrame(bulls)
        Bulls_df.to_csv("The Bulls.csv", index = False)

        print("PIRATES")
        print(Pirates_df)
        print("YANKEES")
        print(Yankees_df)
        print("BULLS")
        print(Bulls_df)

class Encapsulation_error_handling:
    def __init__(self, age, record):
        self.__age = age
        self.name = record["Name"]

    def error_getter(self):
        return f"Student {self.name} is out of range for classification.(Age: {self.__age})"

system = Student_sorting("SIS.csv", data)
system.age_sorting()
```
