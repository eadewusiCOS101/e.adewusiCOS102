```python
from abc import ABC, abstractmethod

class coup_de_ecriva(ABC):
    @abstractmethod
    def Fan_page(self):
        pass

class FC_Cirok(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to FC Cirok"
    
class Madiba_FC(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to Madiba FC"
    
class Blue_Jay_FC(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to Blue Jay FC"
    
class TSG_Walker(coup_de_ecriva):
    def Fan_page(self):
        return "Welcome to TSG Walker"
    
print("Welcome!\nLet's get you settled into your club.")
club_choice = input("Which club are you a fan of?\n1. FC Cirok\n2. Madiba FC\n3. Blue Jay FC\n4. TSG Walker\nEnter the number corresponding to your choice: ")
if club_choice == "1":
    club = FC_Cirok()
    print(club.Fan_page())
elif club_choice == "2":
    club = Madiba_FC()
    print(club.Fan_page())
elif club_choice == "3":
    club = Blue_Jay_FC()
    print(club.Fan_page())
elif club_choice == "4":
    club = TSG_Walker()
    print(club.Fan_page())
else:
    print("Invalid choice. Please enter a number corresponding to your choice.")
```
