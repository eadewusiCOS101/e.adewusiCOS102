```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix
```

```python
AB_Data = pd.read_csv("C:/Users/Guest User/Vproject/e.adewusiCOS102/week-4/Project-1/AB_NYC_2019.csv")
AB_Data = AB_Data.dropna(subset=["name", "host_name","last_review"])

str_columns = ["name", "host_name", "neighbourhood_group", "neighbourhood", "room_type", "last_review"]
for column in str_columns:
    AB_Data.drop(column, axis=1, inplace=True)
AB_Data.info()
```

```python
x = AB_Data.drop(columns = ["price"])
y = AB_Data["price"]
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
dt = DecisionTreeClassifier()
dt.fit(x_train, y_train)

dt_predictions = dt.predict(x_test)
print(accuracy_score(y_test, dt_predictions))
print(precision_score(y_test, dt_predictions, average='weighted'))
print(recall_score(y_test, dt_predictions, average='weighted'))
print(confusion_matrix(y_test, dt_predictions))
```

```python
Kn = KNeighborsClassifier()
Kn.fit(x_train, y_train)

Kn_predictions = dt.predict(x_test)
print(accuracy_score(y_test, Kn_predictions))
print(precision_score(y_test, Kn_predictions, average='weighted'))
print(recall_score(y_test, Kn_predictions, average='weighted'))
print(confusion_matrix(y_test, Kn_predictions))
```
