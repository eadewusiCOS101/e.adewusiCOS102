```python
def int_input(prompt):
    while True:
        try:
            a = int(input(prompt))
            return a
        except ValueError:
            print("Invalid input.")

print("This is the ATR evaluation script.\n")
while True:

    experience = int_input("How many years of experience do you have in this industrial feild?")
    age = int_input("What is your current age?\n")
    if age < 17 or age > 60 or experience > 45:
        print("Invalid entries for evaluation, retry.")
    else:
        if experience > 25 and age >= 55:
            print("Your ATR is ₦5,600,000.00")
        elif experience > 20 and age >= 45:
            print("Your ATR is ₦4,480,000.00")
        elif experience > 10 and age >= 35:
            print("Your ATR is ₦1,500,000.00")
        elif experience < 10 and age < 35:
            print("Your ATR is ₦550,000.00")
        elif (age - experience) < 17:
            print("Invalid inputs for evaluation, retry.")
        else:
            print("Indeterminate ATR. Locate a taxation proffessional within the next 72 hours.")
    print("")

    quit_opt = input("\nWould you like to end the program? Y/N")
    if quit_opt.lower() == "y":
        print("Ending the program.")
        break
```
