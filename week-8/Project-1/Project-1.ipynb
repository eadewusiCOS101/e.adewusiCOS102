```python
import pandas as pd
zenith_employee_file = pd.read_csv("Zenith employees.csv")
print(zenith_employee_file)
```

```python
class zenith:
    def mutual_services(self):
        return ["Lines of credit","Investment management and accounts","Insurance"]
    def unique_services(self):
        return[]
    def display_services(self):
        mutuals = self.mutual_services()
        uniques = self.unique_services()
        services = mutuals + uniques
        return services

class retail_banking(zenith):
    def unique_services(self):
        return ["Retirement and education accounts","loans and mortgagges","Checking and saving"]  

class global_banking(zenith):
    def mutual_services(self):
        return []
    def unique_services(self):
        return ["Multi-currency management services and products",
"Foreign currency accounts",
"Foreign currency credit cards",
"Transborder advisory services",
"Liquidity management"]   
    
class commercial_banking(zenith):
    def unique_services(self):
        return ["Advisory services"] 

```

```python
import tkinter as tk
from tkinter import messagebox as mesgbx
from tkinter import scrolledtext as scroltxt

def check_info ():
    name = name_entry.get().title()
    dept = dept_entry.get().title()
    if dept == "Retail Banking":
        employee = retail_banking()
    elif dept == "Commercial Banking":
        employee = commercial_banking()
    elif dept == "Global Banking":
        employee = global_banking()
    else:
        mesgbx.showerror("Unable to identify department","Sorry, we are having trouble identifying your department\nPlease try again")
    services = employee.display_services()
    result_panel.delete(1.0, tk.END)
    result_panel.insert(tk.END, f"Services provided by the {dept} sector:\n")
    for service in services:
        result_panel.insert(tk.END, f"- {service}\n")

def login():
    name = name_entry.get()
    dept = dept_entry.get()
    name_matching = zenith_employee_file[zenith_employee_file["Full Name"].str.title() == name.title()]
    if name_matching.empty:
        mesgbx.showerror("Error","Sorry, we couldn't find your name in our records. Please try again.")
    else:
        employee_dept = name_matching.iloc[0]["Department"]
        if employee_dept.lower() == dept.lower():
            mesgbx.showinfo("Welcome!",f"Welcome back {name}")
            check_info()
        else:
            mesgbx.showerror("Invalid department", "Sorry, there seems to be a mismatch regarding your department.\nTry again please.")

def clear():
    name_entry.delete(0, tk.END)
    dept_entry.delete(0, tk.END)
    result_panel.delete(1.0, tk.END)

root = tk.Tk()
root.title("Zenith Bank Services")
root.geometry("500x300")

welcome_label = tk.Label(root, text = "Welcome to Zenith bank service review", font = ("Helvetica", 18, "bold"), fg = "red")
welcome_label.pack(pady = 20)

prompt_label = tk.Label(root, text = "Please kindly enter your name.", font = ("Arial", 14))
prompt_label.pack(pady = 10)
name_entry = tk.Entry(root)
name_entry.pack(pady = 5)

prompt_label2 = tk.Label(root, text = "Please kindly enter your department.", font = ("Arial", 14))
prompt_label2.pack(pady = 10)
dept_entry = tk.Entry(root)
dept_entry.pack(pady = 5)

verification_button = tk.Button(root, text = "Login",font = ("Arial", 14),bg = "red",fg = "white", command = login)
verification_button.pack()

clear_button = tk.Button(root, text = "Clear all info", font = ("Arial", 14), bg = "grey", fg = "red", command = clear)
clear_button.pack()

result_panel = scroltxt.ScrolledText(root, width = 50, height = 20)
result_panel.pack(pady = 10)

root.mainloop()
```
