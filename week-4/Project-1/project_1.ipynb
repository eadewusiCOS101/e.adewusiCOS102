```python
import pandas as pd
import matplotlib as mtl
import seaborn as sns

data_in_use = pd.read_csv("AB_NYC_2019.csv")

data_in_use["reviews_per_month"] = data_in_use["reviews_per_month"].fillna(data_in_use["reviews_per_month"].mean())
data_in_use.drop(columns="last_review", inplace=True)
data_in_use.dropna(inplace = True)

print("Data information")
print(data_in_use.info())
print(data_in_use.isna().sum())
```

```python
print("Price histogram.")
sns.histplot(data_in_use["price"], bins=100, kde=True)
```

```python
print("Numeric heatplot")
sns.heatmap(data_in_use.corr(numeric_only=True), annot=True, cmap="coolwarm")
```

```python
print("N.O.R to price scatterplot")
sns.scatterplot(x ="number_of_reviews", y = "price", data = data_in_use)
```

The data in terms of prices seems to be positvely skewed to the left.

Correlation between numeric values is perceived to be low on an average.

The data indicates that most airbnb users review lower cost airbnbs as compared to more expensive options.
